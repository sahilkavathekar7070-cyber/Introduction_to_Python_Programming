Define String and perform various operations

```python
str_1="Hello"
str_2="World"
```

Checking datatype of string

```python
print("Datatype of String is :",type(str_1))
```

Checking Length of String

```python
print("Length of String is :",len(str_2))
```

Concatenation

```python
Greeting=str_1 + " " + str_2
print(Greeting)
```

Methods of String

```python
print(Greeting.upper())
print(Greeting.lower())
print(Greeting.split())
print(Greeting.replace("World","Python"))
```
