Concept of pass, continue and break statements in Python

```python
# pass  pass statement

for i in range(5) :
    if i == 0 :
        pass #This will do nothing and continue to the next iteration
    print(i)
```

```python
# continue statement

for i in range(5) :
    if i == 2 :
        continue # This will skip the rest of the code in the loop for this iteration
    print(i)
```

```python
# break statement

for i in range(5) :
    if i == 2 :
        break  # This will exit the loop completely
    print(i)
```

Another examples

```python
for i in range(1,8) :
    if i % 2 == 0 :
        continue
    print(i)
```

```python
for i in range(1,10) :
    if i % 3 == 0 :
        break
    print(i)
```

```python
for i in range(1,5) :
    if i ==2 :
        continue
    elif i == 4 :
        break
    print(i)
```

Write a program to check whether a given number is prime or not.

```python
def prime():
    num = int(input("Enter a number: "))
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                print(num, "is not a prime number")
                break
        else:
            print(num, "is a prime number")
    else:
        print(num, "is not a prime number")
prime()
```

Write a program to print all the prime numbers between 1 to 100

```python
def primerange():
    for s in  range(1,101):
        if s > 1 :
            for i in range(2,s):
                if(s %i)==0:
                    break
            else :
                print(s)
primerange()
```

Write a program to check armstrong number or not

```python
def arm():
    num=int(input("Enter number :"))
    order=len(str(num))
    sum=0
    temp=num
    while temp > 0 :
        digit=temp % 10
        sum += digit ** order 
        temp //= 10
    if num == sum :
        print(num,"is an armstrong number.")
    else :
        print(num,"is not an armstrong number.")
arm()
```

Write a program to print the fibonacci series

```python
def fibonacci():
    n=int(input("Enter the number of terms :"))
    a,b=0,1
    for i in range(n):
        print(a,end=' ')
        a, b = b, a + b
fibonacci()
```

Write a program to find factorial of a given number

```python
def fact():
    num=int(input("Enter a number :"))
    factorial=1
    if num < 0 :
        print("Factorial does not exist for negative numbers")
    elif num == 0:
        print("Factorial of 0 is 1")
    else :
        for i in range(1,num + 1):
            factorial *=i
        print("The factorial of",num,"is",factorial)
fact()
```

Write a program to find the largest and smallest number in a list

```python
numbers=[5,6,4,9,2,1]
largest = max(numbers)
smallest = min(numbers)
print("Largest number :",largest)
print("Smallest number :",smallest)
```

Write a program to find the second largest number in list

```python
lst = [4,3,6,7,8]
large = max(lst)
second_largest = max(num for num in lst if num != large)
print("Second largest number :",second_largest)
```
