Concept of Indexing and Slicing in python with string

```python
#Indexing

str="Hello, Welcome to the world of python programming"
print(str[0])
print(str[-1])
print(str[12])
```

```python
#Slicing

''' 
string[start:stop:step]
- start: The starting index of the slice (inclusive).
- stop: The ending index of the slice (exclusive).
- step: The step size or the interval between characters in the slice (optional).
'''

print(str[0:14])
print(str[7:28:])
print(str[ : :2])# EVERY SECOND CHARACTER
print(str[ : :-1])#REVERSE
```

Boolean Logic and Comparison Operators

```python
a=10
b=20

print(a>b)
print(a<b)
print(a==b)
print(a!=b)
print(a<=b)
print(a>=b)
```

```python
#Using and logical operator

print((a>5) and (b>15))
print((a>15) and (b>15))
```

```python
#Using or operator

print((a>5) or (b>15))
print((a>15) or (b>15))
print((a>15) or (b>25))
```

Difference between == and = operator

```python
'''
"=" This is Assignment operator, used to assign a value to a variable.

"==" This is Equality operator, used to compare two values are equal or not,
    It return TRUE if values are equal,otherwise FALSE.
'''

x=100
y=200

print(x=y) #ERROR
print(x==y) #FALSE
```
